```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import yfinance as yf
from scipy.optimize import minimize
```

```python
# Downloading stock prices

tickers = ["AAPL", "MSFT", "GOOGL", "AMZN"]
raw_data = yf.download(tickers, start="2020-01-01", auto_adjust = True, progress = False)
prices = raw_data["Close"]
print("Here are the last 5 days of prices we downloaded:")
print(prices.tail())
print()
```

```python
# Turn prices into daily returns

daily_returns = prices.pct_change()
daily_returns.dropna()
print("Here are the last 5 days of daily returns:")
print(daily_returns.tail())
print()
```

```python
# Summary of each stock's behaviours
# Looking at each stock's daily return average and its covariance
# Need to convert from daily numbers to annual numbers
trading_days_per_year = 252
average_daily_return = daily_returns.mean()
mu = [average_daily_return[t] * trading_days_per_year for t in tickers]
print("Annualised average return for each stock:")
for i in range(len(tickers)):
  print(f"{tickers[i]}: {mu[i]:.2%}")
print()

cov_table = daily_returns.cov() * trading_days_per_year

n_stocks = len(tickers)
cov_matrix = []
for i in range(n_stocks):
  row = []
  for j in range(n_stocks):
    row.append(cov_table.iloc[i, j])
  cov_matrix.append(row)

print("Covariance matrix: (how much each stock moves together):")
print(cov_table.round(4))
print()
```

```python
# Scoring any given portfolio

def portfolio_return_and_risk(weights, mu, cov_matrix):
  n = len(weights)
  expected_return = 0
  for i in range(n):
    expected_return = expected_return + weights[i] * mu[i]
  variance = 0
  for i in range(n):
    for j in range(n):
      variance = variance + weights[i] * weights[j] * cov_matrix[i][j]
  volatility = variance ** 0.5
  return expected_return, volatility

equal_weights = [0.25, 0.25, 0.25, 0.25]
ret, vol = portfolio_return_and_risk(equal_weights, mu, cov_matrix)
print(f"Equal-weighted portfolio: expected return = {ret:.2%}, volatility = {vol:.2%}")
print()
```

```python
risk_free_rate = 0.052 # Current UK bond yield
number_of_random_portfolios = 5000

all_returns = []
all_volatilities = []
all_sharpe_ratios = []
all_weights_choices = []

for portfolio_number in range(number_of_random_portfolios):
  random_numbers = [np.random.random() for _ in range(n_stocks)]
  total = sum(random_numbers)
  weights = [r / total for r in random_numbers]
  ret, vol = portfolio_return_and_risk(weights, mu, cov_matrix)

  sharpe_ratio = (ret - risk_free_rate) / vol

  all_returns.append(ret)
  all_volatilities.append(vol)
  all_sharpe_ratios.append(sharpe_ratio)
  all_weights_choices.append(weights)
best_sharpe_so_far = max(all_sharpe_ratios)
best_sharpe_index = all_sharpe_ratios.index(best_sharpe_so_far)
best_weights_random_search = all_weights_choices[best_sharpe_index]

print(f"Best portfolio found by trying {number_of_random_portfolios} random combinations:")
for i in range(n_stocks):
  print(f"{tickers[i]}: {best_weights_random_search[i]:.1%}")
print(f"Return: {all_returns[best_sharpe_index]:.2%}",
      f"Volatility: {all_volatilities[best_sharpe_index]:.2%}",
      f"Sharpe ratio: {best_sharpe_so_far: .3f}")
print()
```

```python
# Finding the exact best answer

def negative_sharpe_ratio(weights_array, mu, cov_matrix, risk_free_rate):
  weights = list(weights_array)
  ret, vol = portfolio_return_and_risk(weights, mu, cov_matrix)
  sharpe = (ret - risk_free_rate) / vol
  return - sharpe

rules = {"type": "eq", "fun": lambda x: np.sum(x)-1}
bounds = [(0,1) for _ in range(n_stocks)]

starting_guess = [1/n_stocks] * n_stocks
result = minimize(
    negative_sharpe_ratio,
    starting_guess,
    args = (mu, cov_matrix, risk_free_rate),
    method = "SLSQP",
    constraints = rules
)
best_weights_exact = result.x
ret_exact, vol_exact = portfolio_return_and_risk(list(best_weights_exact), mu, cov_matrix)
sharpe_exact = (ret_exact - risk_free_rate) / vol_exact

print(f"Return: {ret_exact:.2%}, Volatility: {vol_exact:.2%}, Sharpe ratio: {sharpe_exact:.3f}")
print()
print(f"(Compare: our best random guess only reached Sharpe ratio: {best_sharpe_so_far:.3f}"
      f" the exact tool found something slightly better.)")
```

```python
# Drawing a chart of everything found
fig, ax = plt.subplots(figsize = (10, 6))
scatter = ax.scatter(
    all_volatilities,
    all_returns,
    c = all_sharpe_ratios,
    cmap = "magma",
    s = 10,
    alpha = 0.8
)

ax.set_xlabel("Volatility (risk)")
ax.set_ylabel("Expected Return")
ax.set_title(f"{number_of_random_portfolios} random portfolios")
plt.colorbar(scatter, label = "Sharpe Ratio")
plt.show

fig.tight_layout()
plt.savefig("random_portfolios.png", dpi = 300)
```
